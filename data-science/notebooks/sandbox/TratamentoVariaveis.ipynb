```python
# ==========================================================
# IMPORTAÇÃO DAS BIBLIOTECAS
# ==========================================================

import pandas as pd
import numpy as np
```

```python
# ============================================================
# LEITURA DOS DATASETS (colab)
# ============================================================

usuarios = pd.read_csv(f"{BASE}/processed/usuarios_tratado.csv")
transacoes = pd.read_csv(f"{BASE}/processed/transacoes_tratado.csv")
```

```python
# ============================================================
# LEITURA DOS DATASETS (DRIVE)
# ============================================================
from google.colab import drive
drive.mount('/content/drive')
BASE = "/content/drive/MyDrive/Hackatona G9 Finance AI/data"
```

```python
# ==========================================================
# CONVERSÃO DE TIPOS
# ==========================================================

transacoes["data_transacao"] = pd.to_datetime(
    transacoes["data_transacao"],
    errors="coerce"
)
```

```python
# ==========================================================
# PADRONIZAÇÃO DOS CAMPOS TEXTUAIS
# ==========================================================

texto_usuarios = [

    "sexo",

    "estado_civil",

    "cidade",

    "estado",

    "profissao",

    "escolaridade",

    "frequencia_poupanca",

    "perfil_financeiro"

]

for coluna in texto_usuarios:

    usuarios[coluna] = (

        usuarios[coluna]

        .astype(str)

        .str.strip()

        .str.replace(r"\s+"," ",regex=True)

    )
```

```python
texto_transacoes = [

    "descricao_transacao",

    "categoria",

    "tipo_transacao",

    "forma_pagamento"

]

for coluna in texto_transacoes:

    transacoes[coluna] = (

        transacoes[coluna]

        .astype(str)

        .str.strip()

        .str.replace(r"\s+"," ",regex=True)

    )
```

```python
# ==========================================================
# VARIÁVEIS BINÁRIAS
# ==========================================================

mapa = {

    "Sim":1,

    "Não":0

}
```

```python
binarios_usuario = [

    "possui_financiamento",

    "possui_imovel",

    "possui_veiculo"

]

for coluna in binarios_usuario:

    usuarios[coluna] = usuarios[coluna].map(mapa)
```

```python
binarios_transacoes = [

    "parcelado",

    "transacao_recorrente"

]

for coluna in binarios_transacoes:

    transacoes[coluna] = transacoes[coluna].map(mapa)
```

```python
# ==========================================================
# VALIDAÇÃO DOS TIPOS
# ==========================================================

print("USUÁRIOS")

usuarios.info()

print("\n")

print("TRANSAÇÕES")

transacoes.info()
```

```python
# ==========================================================
# INTEGRIDADE REFERENCIAL
# ==========================================================

usuarios_orfaos = (

    ~transacoes["usuario_id"]

    .isin(

        usuarios["usuario_id"]

    )

).sum()

print(

f"Usuários inexistentes nas transações: {usuarios_orfaos}"

)
```

```python
# ==========================================================
# ORDENAÇÃO
# ==========================================================

usuarios = usuarios.sort_values(

    "usuario_id"

).reset_index(drop=True)

transacoes = transacoes.sort_values(

    ["usuario_id","data_transacao"]

).reset_index(drop=True)
```
